**1. What is Linux?**
  - Linux is a Operating System. It is an Open-source OS, which means it's code is available to the public allowing anyone to modify it. It consists of Kernel (The core engine), a shell (command interface), and various distributions like Ubuntu, Fedora, etc. It controls the core hardware and software part of your system.   

**2. What is the difference between Hard Link & Soft Link?**
  - A Hard Link is a direct copy of a file's "ID" (inode). It points directly to the data on the disk. If you delete the original file, the hard link still works because it is an identical twin.
  - A Soft Link (Symlink) is just a shortcut or a "path" pointing to a filename. If you delete the original file, the soft link breaks because the path it points to is now empty.**bold text**

**3. What is a Kernel in linux?**
  - The Kernal is the core of any operating system, including Linux. It acts as the bridge between your computer's hardware and your software. The kernel is responsible for managing critical tasks like memory, processes, and devices, ensuring everything runs smoothly. It is the first thing that boots up.

**5. What is the ‘grep’ command used for in Linux?**
  - **grep** - Global Regular Expression Print, is essentially a searching tool which goes through files or text for a specific pattern and prints out every line that contains a match.
  - For Example: If you want to search for a word in a file, you will use the command:
  ```
  grep "error" logfile.txt
  ```
