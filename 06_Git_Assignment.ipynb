### **5. Tasks to Be Performed:**

1. Create a Git Flow workflow architecture on Git
- git init
- git add .
- git commit -m "Initial commit"
- git branch develop
- git checkout develop


2. Create all the required branches

Feature branch
- git checkout develop
- git checkout -b feature/login-system

Release branch
- git checkout develop
- git checkout -b release/v1.0

Hotfix branch (created later from master)
- git checkout master
- git checkout -b hotfix/urgent-fix

3. starting from e feature branch, push the branch to the master, following the architecture

feature → develop → release → master

Step A — finish feature into develop
- git checkout feature/login-system
- git add .
- git commit -m "Add login feature"
- git checkout develop
- git merge feature/login-system
- git push origin develop

Step B — create release and merge to master
- git checkout -b release/v1.0 develop
- git push origin release/v1.0

Finalize release:

- git checkout master
- git merge release/v1.0
- git tag v1.0
- git push origin master --tags


Back-merge into develop:

- git checkout develop
- git merge release/v1.0
- git push origin develop


4. Push an urgent.txt on master using hotfix

Create hotfix from master
- git checkout master
- git checkout -b hotfix/urgent-fix

Add urgent file
- echo "URGENT FIX" > urgent.txt
- git add urgent.txt
- git commit -m "Hotfix: add urgent.txt"

Merge hotfix into master
- git checkout master
- git merge hotfix/urgent-fix
- git push origin master


Tag the hotfix:

- git tag v1.0.1
- git push origin v1.0.1

Merge hotfix back into develop (critical step people forget)
- git checkout develop
- git merge hotfix/urgent-fix
- git push origin develop
